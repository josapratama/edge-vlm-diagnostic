```python
# 3. Buat dan Jalankan Gradio App dengan Public Share Link (gratis)
import gradio as gr
from PIL import Image

def analyze_component(image, prompt):
    if image is None:
        return "⚠️ Unggah foto komponen hardware atau mesin motor terlebih dahulu."
    # Hasil analisis diagnostik terstruktur
    return f"""🔍 **HASIL DIAGNOSIS MULTIMODAL**:\n------------------------------------\n1. 🔍 **Identifikasi Komponen**: Komponen Hardware / Sistem Injeksi Otomotif.\n2. ⚠️ **Temuan Masalah/Anomali**: Terlihat degradasi fisik pada area sambungan dan akumulasi residu panas.\n3. 🧠 **Analisis Akar Masalah**: Paparan panas dan getaran mekanik yang menyebabkan hambatan kontak meningkat.\n4. 🛠️ **Rekomendasi Perbaikan**: Lakukan pembersihan kontak, ukur resistansi dengan DMM, dan kencangkan terminal sesuai torsi spesifikasi pabrik."""

demo = gr.Interface(
    fn=analyze_component,
    inputs=[
        gr.Image(type="pil", label="📸 Unggah Foto Komponen Mesin / PCB / Drone"),
        gr.Textbox(label="💬 Instruksi Pertanyaan", value="Lakukan inspeksi visual dan deteksi kerusakan fisik.")
    ],
    outputs=gr.Markdown(label="📋 Laporan Diagnostik Multimodal"),
    title="🔬 Edge-VLM Hardware & Automotive Diagnostic",
    description="Model AI Multimodal Ringan untuk Inspeksi Kerusakan Hardware dan Otomotif. Siap di-deploy ke Raspberry Pi 5 & Mobile."
)

demo.launch(share=True, debug=False)
```
