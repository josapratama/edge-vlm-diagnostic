```python
# 3. Clone / Siapkan generator dataset diagnostik
import json
import random

HARDWARE_CATEGORIES = [
    ("Throttle Body & TPS Motor Injeksi", "Kerak karbon tebal pada katup kupu-kupu dan soket kabel TPS terkelupas", "Motor brebet saat tarikan awal dan putaran idle tidak stabil"),
    ("Regulator LDO AMS1117 ESP32", "Noda gosong pada permukaan IC regulator daya dan kapasitor retak", "Board mikrokontroler mati total akibat lonjakan tegangan input"),
    ("ESC 4-in-1 Drone FPV", "Solderan dingin (cold solder) pada terminal kabel phase motor brushless", "Daya motor tidak sinkron dan rawan terlepas saat manuver getaran tinggi"),
    ("Koil Pengapian Motor Injeksi", "Isolator karet tutup busi sobek dan kabel tegangan tinggi bocor ke rangka", "Percikan api bocor ke ground mesin saat kondisi lembap / hujan"),
    ("Spull Stator Magnet 3-Phase", "Gulungan tembaga spull hitam terbakar akibat panas berlebih", "Pengisian aki drop di bawah 12.4V sehingga aki cepat drop")
]

dataset_records = []
for idx, (comp, defect, symptom) in enumerate(HARDWARE_CATEGORIES * 100):
    record = {
        "id": f"diag_sample_{idx:05d}",
        "conversations": [
            {"from": "human", "value": f"<image>\nLakukan inspeksi visual menyeluruh pada komponen {comp} ini."},
            {"from": "gpt", "value": f"1. 🔍 **Identifikasi Komponen**: {comp}.\n2. ⚠️ **Temuan Masalah/Anomali**: {defect}.\n3. 🧠 **Analisis Akar Penyebab**: {symptom}.\n4. 🛠️ **Rekomendasi Perbaikan**: Lakukan pembersihan area komponen, ukur kontinuitas tegangan dengan DMM, dan lakukan perbaikan isolasi atau penggantian komponen yang rusak."}
        ],
        "image": f"images/sample_{idx % len(HARDWARE_CATEGORIES)}.jpg"
    }
    dataset_records.append(record)

out_file = f"{WORK_DIR}/data/train_sharegpt.json"
with open(out_file, 'w', encoding='utf-8') as f:
    json.dump(dataset_records, f, indent=2, ensure_ascii=False)

print(f"🎉 Sukses membuat {len(dataset_records)} sampel dataset ShareGPT di: {out_file}")
```
